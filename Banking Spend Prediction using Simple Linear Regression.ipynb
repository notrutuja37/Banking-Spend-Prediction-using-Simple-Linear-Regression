```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

```

```python
# Input data ( Annual Income)
X = np.array([30,40,50,60,70,80,90,100,110,120]).reshape(-1,1)
# Output data ( Annual Spending)
y = np.array([10,12,15,18,20,25,28,30,35,40])
```

```python
# Create model
model = LinearRegression()
```

```python
# Train model
model.fit(X, y)
```

```python
# Display slope and intercept
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
# New values for prediction
test_values = np.array([85, 90, 26, 200, 10]).reshape(-1,1)
```

```python
# Predict values
predictions = model.predict(test_values)
# Print predictions
print("Predictions:", predictions)
```

```python
# Plot actual data
plt.scatter(X, y)
# Plot regression line
plt.plot(X, model.predict(X))
# Show graph
plt.show()
```
